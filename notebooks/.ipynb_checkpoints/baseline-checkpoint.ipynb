## 1. Configuração
Importa as bibliotecas e cria as pastas de saída (`poster/figuras` e `modelo`).

```python
import pandas as pd, sklearn, xgboost, shap
import os, glob
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_predict, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (roc_auc_score, recall_score, precision_score, f1_score, confusion_matrix, classification_report, precision_recall_curve, RocCurveDisplay, PrecisionRecallDisplay, ConfusionMatrixDisplay)
```

## 2. Dados e divisão treino/teste
Leitura da base tratada (149.717 registros, 10 variáveis e o alvo `inadimplente_target`). Cerca de 6,6% dos clientes são inadimplentes, uma base fortemente desbalanceada.

A divisão é feita **uma única vez** (80% treino e 20% teste), estratificada pelo alvo e com `random_state=42`, e todos os modelos usam exatamente os mesmos conjuntos. Isso garante uma comparação justa e evita vazamento de dados.

```python
# Buscar dentro de subpastas
arquivos = glob.glob("../dados/**/dados_inadimplencia_limpos*.csv", recursive=True)
print("Arquivo encontrado:", arquivos)
df = pd.read_csv(arquivos[0])

df = pd.read_csv("../dados/dados_inadimplencia_limpos.csv")
ALVO = "inadimplente_target"

x = df.drop(columns=[ALVO])
y = df[ALVO]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=42)

print(x_train.shape, x_test.shape)
print(y_train.mean(), y_test.mean())
```

## 3. Função de avaliação
Calcula ROC-AUC, Recall, Precision, F1 e a matriz de confusão para cada modelo e guarda os resultados na lista `resultados`.

A acurácia não é usada: um modelo que prevê "ninguém inadimplente" já acertaria cerca de 93,4% dos casos sem detectar nenhum inadimplente.

```python
resultados = []
def avaliar(nome, y_verdadeiro, prob, limiar=0.458):
    pred = (prob >= limiar).astype(int)
    r = {
        "modelo": nome,
        "limiar": round(limiar, 3),
        "roc_auc": roc_auc_score(y_verdadeiro, prob),
        "recall": recall_score(y_verdadeiro, pred),
        "precision": precision_score(y_verdadeiro, pred, zero_division=0),
        "f1": f1_score(y_verdadeiro, pred)
    }

    resultados.append(r)
    print(pd.Series(r))
    print(confusion_matrix(y_verdadeiro, pred))
    return r
```

## 4. Baseline: Regressão Logística
Modelo de referência, com padronização (`StandardScaler`) e `class_weight="balanced"` para lidar com o desbalanceamento. É treinado em duas versões:

- **Sem tratar outliers:** ROC-AUC 0,816 e Recall 0,619 (limiar 0,5).
- **Com corte no percentil 99** (`uso_limite_credito`, `razao_endividamento` e `renda_mensal`), calculado somente no treino: ROC-AUC 0,854 e Recall 0,736.

O tratamento melhora bastante o modelo, porque a Regressão Logística é sensível a valores extremos.

```python
# Baseline sem tratar outliers
lr = Pipeline([
    ("escala", StandardScaler()),
    ("lr", LogisticRegression(class_weight="balanced",max_iter=1000))
])
lr.fit(x_train, y_train)
prob_lr = lr.predict_proba(x_test)[:,1]
avaliar("LogReg (sem tratar outliers)", y_test, prob_lr)
```

```python
# Baseline com clipping no percentil 99 - Apenas no treino
cols_clip = ["uso_limite_credito", "razao_endividamento","renda_mensal"]
limites = x_train[cols_clip].quantile(0.99)

def aplicar_clip(x):
    x=x.copy()
    x[cols_clip] = x[cols_clip].clip(upper=limites, axis=1)
    return x

x_train_c, x_test_c = aplicar_clip(x_train), aplicar_clip(x_test)

lr_c = Pipeline([
    ("escala", StandardScaler()),
    ("lr", LogisticRegression(class_weight="balanced", max_iter=1000)),
])
lr_c.fit(x_train_c, y_train)
proba_lr_c = lr_c.predict_proba(x_test_c)[:, 1]
avaliar("LogReg (clip p99)", y_test, proba_lr_c)

coefs = pd.Series(lr_c.named_steps["lr"].coef_[0], index=x.columns).sort_values()
print(coefs)
```

## 5. Random Forest
Conjunto de árvores de decisão, com `class_weight="balanced_subsample"` e profundidade limitada para evitar decorar o treino. Não precisa de padronização nem de corte de outliers. Resultado: ROC-AUC 0,861.

```python
# Random Forest
rf = RandomForestClassifier(
    n_estimators=300, max_depth=12, min_samples_leaf=20,
    class_weight="balanced_subsample", n_jobs=-1, random_state=42
)
rf.fit(x_train, y_train)
prob_rf = rf.predict_proba(x_test)[:, 1]
avaliar("Random Forest", y_test, prob_rf)
```

## 6. XGBoost
Gradient boosting de árvores. O desbalanceamento é tratado com `scale_pos_weight` (razão entre adimplentes e inadimplentes no treino, cerca de 14). Também não exige padronização nem corte de outliers. Resultado: ROC-AUC 0,865.

```python
# XGBoost
spw = (y_train == 0).sum() / (y_train == 1).sum()
print("scale_pos_weight:", spw)

xgb = XGBClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=4,
    subsample=0.8, colsample_bytree=0.8,
    scale_pos_weight=spw, eval_metric="auc",
    n_jobs=-1, random_state=42
)
xgb.fit(x_train, y_train)
proba_xgb = xgb.predict_proba(x_test)[:, 1]
avaliar("XGBoost", y_test, proba_xgb)
```

## 7. Otimização de hiperparâmetros
`RandomizedSearchCV` (20 combinações) com validação cruzada estratificada de 5 folds, **somente no treino**, maximizando o ROC-AUC.

Melhores parâmetros: `n_estimators=200`, `max_depth=4`, `learning_rate=0.05`, `subsample=0.8`, `colsample_bytree=0.8` e `min_child_weight=10`. ROC-AUC médio na validação cruzada: 0,8646.

O ganho sobre o XGBoost inicial foi pequeno (ROC-AUC 0,865 nos dois), o que indica que a configuração inicial já estava próxima do ideal.

```python
# Otimizar hiperparâmetros
param_dist = {
    "n_estimators": [200, 300, 500],
    "max_depth": [3, 4, 5, 6],
    "learning_rate": [0.02, 0.05, 0.1],
    "subsample": [0.7, 0.8, 1.0],
    "colsample_bytree": [0.7, 0.8, 1.0],
    "min_child_weight": [1, 5, 10],
}
busca = RandomizedSearchCV(
    XGBClassifier(scale_pos_weight=spw, eval_metric="auc", n_jobs=-1, random_state=42),
    param_distributions=param_dist, n_iter=20, scoring="roc_auc",
    cv=StratifiedKFold(5, shuffle=True, random_state=42),
    random_state=42, n_jobs=-1, verbose=1
)
busca.fit(x_train, y_train)
print(busca.best_params_, busca.best_score_)

xgb_otimo = busca.best_estimator_
proba_xgb_o = xgb_otimo.predict_proba(x_test)[:, 1]
avaliar("XGBoost otimizado", y_test, proba_xgb_o)
```

## 8. Escolha do limiar de decisão
O limiar é ajustado separadamente do modelo, para atingir um **Recall alvo de 0,80**. Ele é escolhido com probabilidades de validação cruzada no **treino**, e o teste só é usado para reportar o resultado.

Limiar escolhido: **0,458**. No teste: Recall 0,794 e Precision 0,194, com 26,9% da carteira sinalizada.

A tabela de trade-off abaixo mostra o custo de cada alvo de Recall (0,70 a 0,90).

```python
# Limiar para priozar recall
recall_alvo = 0.80
prob_cv = cross_val_predict(
    xgb_otimo, x_train, y_train,
    cv=StratifiedKFold(5, shuffle=True, random_state=42),
    method="predict_proba", n_jobs=-1
)[:, 1]

prec, rec, thr = precision_recall_curve(y_train, prob_cv)
idx = np.where(rec[:-1] >= recall_alvo)[0][-1]
limiar = thr[idx]
print("Limiar:", limiar, "| recall CV:", rec[idx], "| precision CV:", prec[idx])

avaliar(f"XGBoost otimizado (limiar {limiar:.2f})", y_test, proba_xgb_o, limiar=limiar)
```

```python
for alvo in [0.70, 0.75, 0.80, 0.85, 0.90]:
    idx = np.where(rec[:-1] >= alvo)[0][-1]
    lim = thr[idx]
    pred = (proba_xgb_o >= lim).astype(int)
    print(f"alvo {alvo:.2f} | limiar {lim:.3f} | recall {recall_score(y_test, pred):.3f} | precision {precision_score(y_test, pred):.3f}")
```

## 9. Tabela comparativa
Compara os modelos. O ROC-AUC não depende de limiar e é a base da comparação. Recall, Precision e F1 dos demais modelos estão no limiar 0,5, enquanto o XGBoost final está no limiar 0,458, calibrado para o seu escore.

## 10. Gráficos de desempenho
Curvas ROC e Precisão-Recall dos modelos, comparação de Recall e Precision por modelo e matriz de confusão do modelo final (limiar 0,458). As figuras são salvas em `poster/figuras/` para o pôster.

```python
# Gráfico

fig, ax = plt.subplots(figsize=(6, 5))
for nome, p in [("LogReg (clip p99)", proba_lr_c), ("Random Forest", prob_rf),
                ("XGBoost otimizado", proba_xgb_o)]:
    RocCurveDisplay.from_predictions(y_test, p, name=nome, ax=ax)
ax.plot([0, 1], [0, 1], "k--", alpha=0.4)
ax.set_title("Curva ROC")
plt.savefig("../poster/figuras/roc.png", dpi=300, bbox_inches="tight")
plt.show()

fig, ax = plt.subplots(figsize=(6, 5))
for nome, p in [("LogReg (clip p99)", proba_lr_c), ("Random Forest", prob_rf),
                ("XGBoost otimizado", proba_xgb_o)]:
    PrecisionRecallDisplay.from_predictions(y_test, p, name=nome, ax=ax)
ax.set_title("Curva Precisão-Recall")
plt.savefig("../poster/figuras/pr.png", dpi=300, bbox_inches="tight")
plt.show()

pred_final = (proba_xgb_o >= limiar).astype(int)
ConfusionMatrixDisplay.from_predictions(
    y_test, pred_final, display_labels=["Adimplente", "Inadimplente"], cmap="Blues"
)
plt.title("Matriz de confusão (modelo final)")
plt.savefig("../poster/figuras/matriz_confusao.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Gráfico - Recall/Precisão
modelos = [("LogReg (clip p99)", proba_lr_c, 0.5),
           ("Random Forest", prob_rf, 0.5),
           ("XGBoost otimizado", proba_xgb_o, limiar)]

nomes = [m[0] for m in modelos]
recalls, precisoes = [], []
for nome, p, lim in modelos:
    pred = (p >= lim).astype(int)
    recalls.append(recall_score(y_test, pred))
    precisoes.append(precision_score(y_test, pred))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Esquerda: Recall
b1 = ax1.bar(nomes, recalls, color=["#6c8ebf", "#82b366", "#d6b656"])
ax1.set_title("Recall (Sensibilidade)")
ax1.set_ylim(0, 1)
ax1.bar_label(b1, fmt="%.3f", padding=3)

# Direita: Precision
b2 = ax2.bar(nomes, precisoes, color=["#6c8ebf", "#82b366", "#d6b656"])
ax2.set_title("Precisão")
ax2.set_ylim(0, 1)
ax2.axhline(y_test.mean(), color="gray", linestyle=":", label="Taxa base (6,6%)")
ax2.bar_label(b2, fmt="%.3f", padding=3)
ax2.legend()

for ax in (ax1, ax2):
    ax.tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.savefig("../poster/figuras/recall_precisao_barras.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 11. Explicabilidade (SHAP)
Importância das variáveis no XGBoost final, calculada sobre uma amostra de 3.000 clientes do teste. Cada barra mostra o impacto médio da variável no escore de risco (média do valor absoluto do SHAP, em log-odds).

Principais variáveis: `uso_limite_credito` (32,6%), `qtd_atrasos_30_59_dias` (15,6%) e `qtd_atrasos_90_dias` (12,7%). O gráfico mostra a **importância**, e não a direção do efeito nem causalidade.

```python
# SHAP - Explicabilidade
amostra = x_test.sample(3000, random_state=42)
explainer = shap.TreeExplainer(xgb_otimo)
shap_values = explainer.shap_values(amostra)
#shap.summary_plot(shap_values, amostra, show=False)
#plt.savefig("../poster/figuras/shap_resumo.png", dpi=300, bbox_inches="tight")
#plt.show()

ordem = np.abs(shap_values).mean(axis=0).argsort()[::-1]
dados_box = [shap_values[:, i] for i in ordem]
nomes = [amostra.columns[i] for i in ordem]

fig, ax = plt.subplots(figsize=(8, 6))
ax.boxplot(dados_box, vert=False, showfliers=False)
ax.set_yticks(range(1, len(nomes) + 1))
ax.set_yticklabels(nomes)
ax.invert_yaxis()                      # variável mais importante no topo
ax.axvline(0, color="gray", linestyle="--", alpha=0.6)
ax.set_xlabel("Valor SHAP (impacto na saída do modelo)")
ax.set_title("Distribuição do impacto de cada variável")
plt.savefig("../poster/figuras/shap_boxplot.png", dpi=300, bbox_inches="tight")
plt.show()

shap.summary_plot(shap_values, amostra, plot_type="bar", show=False)
plt.xlabel("Impacto médio no escore de risco\n(média do |valor SHAP|)")
plt.savefig("../poster/figuras/shap_importancia.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 12. Análises complementares
Tabelas simples para interpretar o comportamento do modelo:

- **Uso do limite de crédito:** a inadimplência real cresce de 2,2% (até 25% do limite) para 35,2% (acima de 100%).
- **Faixa etária:** clientes mais jovens inadimplem mais e são mais sinalizados. A precisão dos alertas é parecida entre as faixas (0,18 a 0,21), mas o Recall cai de 0,894 (18-29) para 0,479 (70+, com poucos casos).

```python
LIMIAR = limiar
pred_final = (proba_xgb_o >= LIMIAR).astype(int)
print("Limiar:", LIMIAR)
```

```python
teste = x_test.copy()
teste["real"] = y_test.values
teste["sinalizado"] = (proba_xgb_o >= LIMIAR).astype(int)
teste["faixa"] = pd.cut(teste["idade"], bins=[17, 29, 39, 49, 59, 69, 100],
                        labels=["18-29", "30-39", "40-49", "50-59", "60-69", "70+"])

res = teste.groupby("faixa", observed=True).apply(lambda g: pd.Series({
    "clientes": len(g),
    "inadimplencia_real": g["real"].mean(),
    "sinalizados": g["sinalizado"].mean(),
    "recall": g.loc[g["real"] == 1, "sinalizado"].mean(),
    "precision": g.loc[g["sinalizado"] == 1, "real"].mean(),
}))
res.round(3)
```

```python
teste["faixa_uso"] = pd.cut(teste["uso_limite_credito"],
                            bins=[-0.01, 0.25, 0.5, 0.75, 1.0, 1e9],
                            labels=["até 25%", "25-50%", "50-75%", "75-100%", "acima de 100%"])
teste.groupby("faixa_uso", observed=True)["real"].agg(["count", "mean"]).round(3)
```

Salva em `modelo/modelo_previnad.pkl` um dicionário com o XGBoost otimizado, a lista das 10 colunas na ordem de entrada e o limiar (0,458). Um cliente é sinalizado quando o escore de risco é maior ou igual ao limiar.

O XGBoost não usa padronização nem corte de outliers. O arquivo também é testado com `joblib.load`. As tabelas finais são salvas em `poster/`.

```python
artefato = {
    "modelo": xgb_otimo,
    "colunas": list(x_train.columns),
    "limiar": float(LIMIAR),
}
joblib.dump(artefato, "../modelo/modelo_previnad.pkl")

# teste: carregar e prever uma linha
art = joblib.load("../modelo/modelo_previnad.pkl")
exemplo = x_test[art["colunas"]].iloc[[0]]
escore = art["modelo"].predict_proba(exemplo)[0, 1]
print(art["colunas"], art["limiar"], escore, escore >= art["limiar"])

# arquivos de apoio para o pôster
pd.Series(busca.best_params_).to_csv("../poster/melhores_hiperparametros.csv", header=["valor"])
pd.DataFrame(confusion_matrix(y_test, pred_final),
             index=["Real: adimplente", "Real: inadimplente"],
             columns=["Previsto: adimplente", "Previsto: inadimplente"]
            ).to_csv("../poster/matriz_confusao.csv")

print("Tamanho do .pkl (MB):", round(os.path.getsize("../modelo/modelo_previnad.pkl") / 1e6, 2))
print("scikit-learn", sklearn.__version__, "| xgboost", xgboost.__version__,
      "| shap", shap.__version__, "| joblib", joblib.__version__, "| pandas", pd.__version__)
```

```python
# Salvar resultados no .csv
# métricas do limiar final
final = pd.DataFrame([{
    "modelo": "XGBoost otimizado (limiar final)",
    "limiar": LIMIAR,
    "recall": recall_score(y_test, pred_final),
    "precision": precision_score(y_test, pred_final),
    "f1": f1_score(y_test, pred_final),
}])
final.round(3).to_csv("../poster/metricas_limiar_final.csv", index=False)

# matriz de confusão
cm = confusion_matrix(y_test, pred_final)
pd.DataFrame(cm,
             index=["Real: adimplente", "Real: inadimplente"],
             columns=["Previsto: adimplente", "Previsto: inadimplente"]
            ).to_csv("../poster/matriz_confusao.csv")
```

## 14. Conclusões e limitações

- **Modelo final:** XGBoost otimizado, limiar 0,458. De 1.976 inadimplentes no teste, detecta 1.569 (Recall 0,794) e sinaliza 8.069 clientes, dos quais 1.569 são inadimplentes (Precision 0,194, cerca de 2,9 vezes a taxa base de 6,6%).
- **Custo da priorização do Recall:** cerca de 9 a 10 falsos alertas por inadimplente adicional detectado.
- **O escore não é uma probabilidade calibrada** (usa `scale_pos_weight`). Na interface, deve ser mostrado como escore de risco.
- **Imputação pela mediana** feita na base inteira pelo Membro 1 (vazamento leve).
- **Associação, não causalidade:** a importância SHAP descreve o peso das variáveis no modelo.
- **Idade:** é a 4ª variável mais influente, e o Recall é menor entre os mais velhos. O escore deve apoiar a revisão humana, e não decidir sozinho a concessão de crédito.
